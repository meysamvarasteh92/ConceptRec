## JSD @20 

```python

import torch
import numpy as np
from scipy.spatial.distance import jensenshannon

# --------------------------------------------------
# 1. Configuration
# --------------------------------------------------
K = 20
MASK_SEEN_ITEMS = True

# --------------------------------------------------
# Model evaluation and cache initialization
# --------------------------------------------------

base_model.eval()
CBM_model.eval()

# Clear cached LightGCN embeddings.
# They may have been restored on CPU from the checkpoint.
base_model.restore_user_e = None
base_model.restore_item_e = None

# Clear cached propagated embeddings in LightGCN-CBM
if hasattr(CBM_model, "invalidate_cache"):
    CBM_model.invalidate_cache()

n_items = dataset.item_num

assert base_model.n_items == CBM_model.n_items == n_items, \
    "The models must use the same item vocabulary."

assert base_model.n_users == CBM_model.n_users, \
    "The models must use the same user vocabulary."

base_device = next(base_model.parameters()).device
cbm_device = next(CBM_model.parameters()).device

base_counts = np.zeros(n_items, dtype=np.int64)
cbm_counts = np.zeros(n_items, dtype=np.int64)

n_test_inputs = 0
n_identical = 0


# --------------------------------------------------
# 2. Compute full-ranking scores
# --------------------------------------------------
def get_scores(model, interaction, device):
    """
    LightGCN uses USER_ID rather than ITEM_SEQ.

    full_sort_predict() computes recommendation scores
    using propagated user and item embeddings.
    """
    batch = interaction.to(device)

    scores = model.full_sort_predict(batch)

    return scores.reshape(len(interaction), n_items).clone()


# --------------------------------------------------
# 3. Generate top-K lists for BOTH models
# --------------------------------------------------
with torch.inference_mode():

    for interaction, history_index, _, _ in test_data:

        # Original LightGCN scores
        base_scores = get_scores(
            base_model, interaction, base_device
        )

        # LightGCN-CBM scores
        cbm_scores = get_scores(
            CBM_model, interaction, cbm_device
        )

        # Remove padding item
        base_scores[:, 0] = -float("inf")
        cbm_scores[:, 0] = -float("inf")

        # --------------------------------------------------
        # Mask previously interacted items
        # --------------------------------------------------
        if MASK_SEEN_ITEMS and history_index is not None:

            # history_index = (batch_user_rows, item_ids)
            history_u, history_i = history_index

            # Mask for original LightGCN
            base_scores[
                history_u.to(base_device),
                history_i.to(base_device)
            ] = -float("inf")

            # Mask for LightGCN-CBM
            cbm_scores[
                history_u.to(cbm_device),
                history_i.to(cbm_device)
            ] = -float("inf")

        # Ensure enough candidate items remain
        if (
            (torch.isfinite(base_scores).sum(dim=1) < K).any()
            or (torch.isfinite(cbm_scores).sum(dim=1) < K).any()
        ):
            raise ValueError(
                f"Some users have fewer than {K} candidate items."
            )

        # --------------------------------------------------
        # Top-K recommendations
        # --------------------------------------------------
        base_topk = torch.topk(
            base_scores, K, dim=1
        ).indices.cpu().numpy()

        cbm_topk = torch.topk(
            cbm_scores, K, dim=1
        ).indices.cpu().numpy()

        # --------------------------------------------------
        # Count item exposures
        # --------------------------------------------------
        base_counts += np.bincount(
            base_topk.ravel(),
            minlength=n_items
        )

        cbm_counts += np.bincount(
            cbm_topk.ravel(),
            minlength=n_items
        )

        n_test_inputs += len(interaction)

        n_identical += np.all(
            base_topk == cbm_topk,
            axis=1
        ).sum()


# --------------------------------------------------
# 4. Convert item counts to probabilities
# --------------------------------------------------
if n_test_inputs == 0:
    raise ValueError("No test users were evaluated.")

# Exclude padding item 0
p = base_counts[1:].astype(float)
q = cbm_counts[1:].astype(float)

p /= p.sum()
q /= q.sum()


# --------------------------------------------------
# 5. Jensen-Shannon Divergence
# --------------------------------------------------
# scipy returns Jensen-Shannon DISTANCE.
# Squaring gives Jensen-Shannon DIVERGENCE.

jsd = jensenshannon(p, q, base=2) ** 2

print(f"Test users: {n_test_inputs}")

print(f"JSD@{K}: {jsd:.6f}")

print(
    f"Identical top-{K} ranked lists: "
    f"{n_identical / n_test_inputs:.2%}"
)

print(
    f"Backbone item coverage: "
    f"{np.mean(base_counts[1:] > 0):.4f}"
)

print(
    f"ConceptRec item coverage: "
    f"{np.mean(cbm_counts[1:] > 0):.4f}"
)
```

```python

import torch
import numpy as np
import torch.nn.functional as F
import rbo


@torch.no_grad()
def get_top_k_recommendations(
    model,
    data_loader,
    k=20,
    exclude_seen=True
):
    """
    Get top-k recommendations for LightGCN or LightGCN_CBM.

    Returns:
        {user_id: LongTensor[k]}

    Uses RecBole history_index to mask previously
    interacted items.
    """

    model.eval()
    device = next(model.parameters()).device
    recs = {}

    # Clear cached original LightGCN embeddings.
    # Prevents CPU/GPU mismatch after loading checkpoints.
    if hasattr(model, "restore_user_e"):
        model.restore_user_e = None
        model.restore_item_e = None

    # Clear LightGCN_CBM propagation cache.
    if hasattr(model, "invalidate_cache"):
        model.invalidate_cache()

    for batch in data_loader:

        if isinstance(batch, (tuple, list)):
            interaction = batch[0]
            history_index = batch[1]
        else:
            interaction = batch
            history_index = None

        interaction = interaction.to(device)

        # LightGCN uses user IDs, not sequences
        user_ids = interaction[model.USER_ID]
        B = len(user_ids)

        # Full-catalog recommendation scores
        scores = model.full_sort_predict(interaction)
        scores = scores.reshape(B, model.n_items).clone()

        # Exclude padding
        scores[:, 0] = float("-inf")

        # Exclude previously interacted items
        if exclude_seen:
            if history_index is None:
                raise ValueError(
                    "history_index is missing. "
                    "Use RecBole's full-sort evaluation loader."
                )

            history_u, history_i = history_index

            scores[
                history_u.to(device),
                history_i.to(device)
            ] = float("-inf")

        # Prevent invalid recommendations
        if (torch.isfinite(scores).sum(dim=1) < k).any():
            raise ValueError(
                f"Some users have fewer than {k} "
                "available recommendation candidates."
            )

        topk = torch.topk(
            scores, k, dim=1
        ).indices

        # Store internal user IDs and recommendations
        for uid, row in zip(
            user_ids.cpu().tolist(),
            topk
        ):
            recs[int(uid)] = row.detach().cpu()

    return recs
base_recs = get_top_k_recommendations(
    model=base_model,
    data_loader=test_data,
    k=20
)

cbm_recs = get_top_k_recommendations(
    model=CBM_model,
    data_loader=test_data,
    k=20
)
```

```python
common_users = sorted(set(base_recs) & set(cbm_recs))

rbo_scores = []

for uid in common_users:

    original = base_recs[uid].tolist()
    reconstructed = cbm_recs[uid].tolist()

    score = rbo.RankingSimilarity(
        original,
        reconstructed
    ).rbo_ext(p=0.9)

    rbo_scores.append(score)

print(f"Users evaluated: {len(common_users)}")
print(f"Mean RBO@20: {np.mean(rbo_scores):.4f}")
```

```python

@torch.no_grad()
def collect_h_hhat(
    cbm_model,
    data_loader,
    device=None
):
    """
    Collect original and reconstructed LightGCN
    user representations.

    h:     propagated LightGCN user embedding
    h_hat: reconstructed CBM user embedding

    Returns:
        H:     [N, 128]
        H_hat: [N, 128]
    """

    if device is None:
        device = next(cbm_model.parameters()).device

    cbm_model.eval()

    # Clear old propagation cache
    if hasattr(cbm_model, "invalidate_cache"):
        cbm_model.invalidate_cache()

    H = []
    H_hat = []

    for batch in data_loader:

        interaction = (
            batch[0]
            if isinstance(batch, (tuple, list))
            else batch
        )

        interaction = interaction.to(device)

        # LightGCN uses USER_ID
        user = interaction[cbm_model.USER_ID]

        # Get original and reconstructed representations
        h, c_hat, z, h_hat = cbm_model.forward(user)

        H.append(h.detach().cpu())
        H_hat.append(h_hat.detach().cpu())

    if not H:
        raise ValueError("No test users were evaluated.")

    return (
        torch.cat(H, dim=0),
        torch.cat(H_hat, dim=0)
    )

def embedding_similarity_report(H, H_hat):
    """
    Compare original and reconstructed user embeddings.

    H:     [N, hidden_size]
    H_hat: [N, hidden_size]
    """

    assert H.shape == H_hat.shape, \
        f"Shape mismatch: {H.shape} vs {H_hat.shape}"

    N, D = H.shape

    H = H.float()
    H_hat = H_hat.float()

    # Per-user cosine similarity
    cos = F.cosine_similarity(
        H,
        H_hat,
        dim=-1
    )

    metrics = {
        "cosine_mean": cos.mean().item(),
        "cosine_median": cos.median().item(),
        "cosine_std": cos.std(unbiased=False).item(),
    }

    print(f"\n{'=' * 60}")
    print(
        f"Embedding similarity: h vs h_hat "
        f"(N={N}, dim={D})"
    )
    print(f"{'=' * 60}")

    print(
        f"Cosine mean:   {metrics['cosine_mean']:.4f}"
    )

    print(
        f"Cosine median: {metrics['cosine_median']:.4f}"
    )

    print(
        f"Cosine std:    {metrics['cosine_std']:.4f}"
    )

    return metrics
```

```python
# Disable steering to measure reconstruction fidelity
CBM_model._interventions = {}
CBM_model.steer_concept_idx = None
CBM_model.steer_deltas = None

H, H_hat = collect_h_hhat(
    CBM_model,
    test_data
)

metrics = embedding_similarity_report(
    H,
    H_hat
)
```

## Zeroing-out top-k neurons

```python
"""Zero-ablation ranking-stability evaluation for a trained LightGCN_CBM.

Requires: torch, numpy, rbo, and a RecBole FullSortEvalDataLoader.
Existing notebook variables: CBM_model and test_data.
"""

import numpy as np
import torch
import rbo


@torch.no_grad()
def _encode_cbm(cbm_model, interaction):
    """Get user-specific predicted concepts and free representation."""
    users = interaction[cbm_model.USER_ID]
    propagated_users, _ = cbm_model._ensure_propagated()
    h = propagated_users[users]
    c_hat = torch.sigmoid(cbm_model.concept_predictor(h))
    z = cbm_model.latent_predictor(h)
    return c_hat, z


@torch.no_grad()
def _recommend(cbm_model, interaction, c_hat, z, k=20,
               seen_items=None, exclude_seen=True):
    """Rank the propagated item embeddings using reconstructed LightGCN user state."""
    _, propagated_items = cbm_model._ensure_propagated()
    h_hat = cbm_model.reconstructor(torch.cat([c_hat, z], dim=-1))
    scores = (h_hat @ propagated_items.T)[0].clone()  # one user

    scores[0] = float('-inf')  # RecBole padding item
    if exclude_seen:
        if seen_items is None:
            raise ValueError(
                'Missing seen_items. For LightGCN, pass per-user items from '
                'the test loader history_index (not ITEM_SEQ).'
            )
        seen = torch.as_tensor(seen_items, dtype=torch.long, device=scores.device)
        seen = seen[(seen > 0) & (seen < scores.numel())]
        if seen.numel():
            scores[seen] = float('-inf')

    if int(torch.isfinite(scores).sum().item()) < k:
        raise ValueError(f'Fewer than {k} unmasked candidates for this user.')

    return torch.topk(scores, k).indices.cpu().tolist()


@torch.no_grad()
def steer_single_user(
    cbm_model, interaction, top_n=3, k=20,
    select_by='largest', seed=42, device=None, verbose=True,
    seen_items=None, exclude_seen=True,
):
    """Ablate concepts for exactly one LightGCN user."""
    if device is None:
        device = next(cbm_model.parameters()).device
    cbm_model.eval()
    interaction = interaction.to(device)
    if interaction[cbm_model.USER_ID].numel() != 1:
        raise ValueError('steer_single_user expects one user.')
    if top_n < 0:
        raise ValueError('top_n must be nonnegative.')

    # Guard against a cache created on CPU before model.to('cuda').
    cached_users = getattr(cbm_model, '_prop_user_e', None)
    cached_items = getattr(cbm_model, '_prop_item_e', None)
    if (cached_users is not None and cached_users.device != device) or (
        cached_items is not None and cached_items.device != device
    ):
        cbm_model.invalidate_cache()

    c_hat, z = _encode_cbm(cbm_model, interaction)
    n = min(top_n, c_hat.shape[1])

    if select_by == 'largest':
        top_idx = torch.topk(c_hat[0], n, largest=True).indices
    elif select_by == 'smallest':
        top_idx = torch.topk(c_hat[0], n, largest=False).indices
    elif select_by == 'random':
        rng = torch.Generator(device='cpu').manual_seed(seed)
        top_idx = torch.randperm(c_hat.shape[1], generator=rng)[:n].to(device)
    else:
        raise ValueError("select_by must be 'largest', 'smallest', or 'random'")

    baseline = _recommend(
        cbm_model, interaction, c_hat, z, k,
        seen_items=seen_items, exclude_seen=exclude_seen,
    )
    c_steered = c_hat.clone()
    c_steered[0, top_idx] = 0.0
    steered = _recommend(
        cbm_model, interaction, c_steered, z, k,
        seen_items=seen_items, exclude_seen=exclude_seen,
    )

    overlap = len(set(baseline) & set(steered))
    changed = len(baseline) - overlap
    rbo_score = rbo.RankingSimilarity(baseline, steered).rbo_ext(p=0.9)
    names = getattr(cbm_model, 'concept_names', None)
    selected = [
        (idx,
         str(names[idx]) if names is not None and idx < len(names) else f'concept_{idx}',
         float(c_hat[0, idx]))
        for idx in top_idx.tolist()
    ]
    if verbose:
        print(f'\nSelection: {select_by}')
        print('Concepts set to zero:')
        for idx, name, value in selected:
            print(f'  [{idx}] {name}: {value:.4f}')
        print(f'RBO: {rbo_score:.4f}')
        print(f'Overlap: {overlap}/{len(baseline)}')
        print(f'Changed: {changed}/{len(baseline)}')

    return {
        'select_by': select_by,
        'selected_concepts': selected,
        'baseline': baseline,
        'steered': steered,
        'rbo': rbo_score,
        'overlap': overlap,
        'changed': changed,
    }


@torch.no_grad()
def steer_population(
    cbm_model, data_loader, top_n=3, k=20,
    select_by='largest', seed=42, max_users=None,
    device=None, exclude_seen=True,
):
    """Evaluate one LightGCN user per row, using full-sort history_index."""
    if device is None:
        device = next(cbm_model.parameters()).device
    cbm_model.eval()
    cbm_model.invalidate_cache()  # propagate on current model device once

    rbo_scores, changed_items = [], []
    n_users = 0

    for batch in data_loader:
        if isinstance(batch, (tuple, list)):
            interaction = batch[0]
            history_index = batch[1] if len(batch) > 1 else None
        else:
            interaction = batch
            history_index = None

        interaction = interaction.to(device)
        batch_size = interaction[cbm_model.USER_ID].numel()
        seen_for_row = [[] for _ in range(batch_size)]

        if exclude_seen:
            if history_index is None:
                raise ValueError(
                    'Expected RecBole FullSortEvalDataLoader history_index. '
                    'Do not mask by ITEM_SEQ for LightGCN.'
                )
            history_rows, history_items = history_index
            # These are BATCH-ROW indices, not global user IDs.
            for row, item in zip(
                history_rows.cpu().tolist(), history_items.cpu().tolist()
            ):
                if 0 <= row < batch_size and item > 0:
                    seen_for_row[row].append(item)

        for i in range(batch_size):
            result = steer_single_user(
                cbm_model,
                interaction[i:i + 1],
                top_n=top_n,
                k=k,
                select_by=select_by,
                seed=seed + n_users,
                device=device,
                verbose=False,
                seen_items=seen_for_row[i],
                exclude_seen=exclude_seen,
            )
            rbo_scores.append(result['rbo'])
            changed_items.append(result['changed'])
            n_users += 1
            if max_users is not None and n_users >= max_users:
                break
        if max_users is not None and n_users >= max_users:
            break

    if n_users == 0:
        raise ValueError('No users evaluated.')
    rbo_scores = np.asarray(rbo_scores)
    changed_items = np.asarray(changed_items)
    print(f'\nSelection: {select_by}')
    print(f'Users: {n_users}')
    print(f'Mean RBO@{k}: {rbo_scores.mean():.4f}')
    print(f'Mean changed items: {changed_items.mean():.2f}/{k}')
    print(f'Users with >=1 changed item: {(changed_items > 0).mean() * 100:.1f}%')
    return {
        'select_by': select_by,
        'rbo': rbo_scores,
        'changed': changed_items,
        'n_users': n_users,
    }



```

## Exact and Overlap

```python

import torch
import numpy as np

# ---------------------------------------
# 1. Configuration
# ---------------------------------------
BACKBONE = "LightGCN"  # SASRec or "LightGCN"
k = 3
MIN_ACTIVE_USERS = 10

SAE_model.eval()
CBM_model.eval()

device = next(SAE_model.parameters()).device

assert next(CBM_model.parameters()).device == device, \
    "SAE and CBM must be on the same device."


# ---------------------------------------
# 2. Extract SAE activations and GT
# ---------------------------------------
@torch.no_grad()
def extract_data(loader, deduplicate=False):

    all_z = []
    all_gt = []
    seen_users = set()

    for batch in loader:

        interaction = (
            batch[0] if isinstance(batch, tuple)
            else batch
        ).to(device)

        if BACKBONE == "SASRec":

            seq = interaction[SAE_model.ITEM_SEQ]
            seq_len = interaction[SAE_model.ITEM_SEQ_LEN]

            h = SAE_model._encode(seq, seq_len)

            gt = CBM_model._lookup_concepts(
                interaction[CBM_model.ITEM_SEQ]
            )

            users = None

        elif BACKBONE == "LightGCN":

            users = interaction[SAE_model.USER_ID].view(-1)

            propagated_users, _ = (
                SAE_model._ensure_propagated()
            )

            h = propagated_users[users]

            gt = CBM_model._lookup_concepts(
                interaction[CBM_model.USER_ID].view(-1)
            )

        else:
            raise ValueError("Unknown backbone")

        # Original SAE activations, no steering
        z = SAE_model.sae.encode(h)

        if deduplicate and users is not None:

            keep = []

            for i, uid in enumerate(users.tolist()):
                if uid not in seen_users:
                    seen_users.add(uid)
                    keep.append(i)

            if not keep:
                continue

            z = z[keep]
            gt = gt[keep]

        all_z.append(z.detach().cpu().numpy())
        all_gt.append(gt.detach().cpu().numpy())

    return (
        np.concatenate(all_z, axis=0),
        np.concatenate(all_gt, axis=0)
    )


# ---------------------------------------
# 3. Training data for neuron alignment
# ---------------------------------------
Z_train, GT_train = extract_data(
    train_data,
    deduplicate=(BACKBONE == "LightGCN")
)

print("Train SAE:", Z_train.shape)
print("Train GT:", GT_train.shape)


# ---------------------------------------
# 4. Pearson correlation
#    SAE neurons vs predefined concepts
# ---------------------------------------
z_mean = Z_train.mean(axis=0)
z_std = Z_train.std(axis=0)

gt_mean = GT_train.mean(axis=0)
gt_std = GT_train.std(axis=0)

eps = 1e-8

X = (Z_train - z_mean) / np.maximum(z_std, eps)
Y = (GT_train - gt_mean) / np.maximum(gt_std, eps)

corr = (X.T @ Y) / len(X)

# Exclude inactive and constant features
active = (Z_train > 0).sum(axis=0)

valid = (
    (active >= MIN_ACTIVE_USERS)
    & (z_std > eps)
)

corr[~valid, :] = -np.inf
corr[:, gt_std <= eps] = -np.inf


# ---------------------------------------
# 5. Map each concept to a distinct SAE
#    neuron using training correlations
# ---------------------------------------
n_concepts = GT_train.shape[1]

mapping = np.full(n_concepts, -1, dtype=int)
used = set()

# Assign the strongest-aligned concepts first
concept_order = np.argsort(
    -np.max(corr, axis=0)
)

for j in concept_order:

    candidates = np.argsort(-corr[:, j])

    for neuron in candidates:

        if (
            neuron not in used
            and np.isfinite(corr[neuron, j])
            and corr[neuron, j] > 0
        ):
            mapping[j] = neuron
            used.add(neuron)
            break

if np.any(mapping < 0):
    missing = np.where(mapping < 0)[0]
    raise ValueError(
        f"No positively aligned SAE neuron for "
        f"concept indices: {missing.tolist()}"
    )

print("Mapped concepts:", len(mapping))
print("Mean alignment correlation:",
      corr[mapping, np.arange(n_concepts)].mean())


# ---------------------------------------
# 6. Extract test activations and GT
# ---------------------------------------
Z_test, gt = extract_data(test_data)

print("Test SAE:", Z_test.shape)
print("Test GT:", gt.shape)


# ---------------------------------------
# 7. Convert mapped neuron activations
#    into concept prediction scores
#
#    One-feature linear calibration
#    fitted on training data only
# ---------------------------------------
selected_corr = corr[
    mapping, np.arange(n_concepts)
]

c_hat = (
    gt_mean[None, :]
    + (
        (Z_test[:, mapping] - z_mean[mapping])
        / np.maximum(z_std[mapping], eps)
    ) * (selected_corr * gt_std)[None, :]
)

c_hat = np.clip(c_hat, 0.0, 1.0)


# ---------------------------------------
# 8. Ordered top-3 accuracy
# ---------------------------------------
pred_top = np.argsort(-c_hat, axis=1)[:, :k]
gt_top = np.argsort(-gt, axis=1)[:, :k]

# Same soft metric as ConceptRec
soft_accuracy = (
    pred_top == gt_top
).mean(axis=1).mean()

# Same strict metric as ConceptRec
strict_accuracy = np.all(
    pred_top == gt_top,
    axis=1
).mean()


# ---------------------------------------
# 9. Additional unordered overlap
# ---------------------------------------
overlap = np.mean([
    len(set(p) & set(g)) / k
    for p, g in zip(pred_top, gt_top)
])


# ---------------------------------------
# 10. Results
# ---------------------------------------
print(f"SAE Soft Accuracy@{k}:   {soft_accuracy:.4f}")
print(f"SAE Strict Accuracy@{k}: {strict_accuracy:.4f}")
print(f"SAE Overlap@{k}:         {overlap:.4f}")
```
